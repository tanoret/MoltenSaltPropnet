```python
import pandas as pd
import re 
import ast as ast
data = pd.read_csv('new_mstdb_janz.csv')
chem = pd.read_excel('Chemical_table.xlsx')
```

```python
chem = chem.dropna(subset =["Symbol"]).copy()
chem["symbol"] = chem["Symbol"].str.strip()
chem.head()

chem_propertiers = [c for c in chem.columns if c!= "Symbol"]
chem_propertiers    
property_maps = {prop: chem.set_index("Symbol")[prop].to_dict() 
                for prop in chem_propertiers}
```

```python
def parse_composition(comp_value):
    """Wandelt die Composition-Spalte von String in dict um."""
    if isinstance(comp_value, dict):
        return comp_value
    if isinstance(comp_value, str):
        try:
            return ast.literal_eval(comp_value)
        except (SyntaxError, ValueError):
            return {}
    return {}
```

```python
def build_property_dict(comp_value, prop_map):
    """
    comp_value: Inhalt von df['Composition'] (String mit dict)
    prop_map:   Dict Symbol -> property_wert (z.B. AtomicMass(Element))
    Rückgabe:   Dict {Symbol: Wert} nur für Symbole, die in Composition UND in prop_map sind
    """
    comp = parse_composition(comp_value)
    result = {}
    for symbol in comp.keys():
        if symbol in prop_map:
            result[symbol] = prop_map[symbol]
        # Symbole, die nicht in Chemical_table sind, werden ignoriert
    return result
```

```python
for prop_name, prop_map in property_maps.items():
    # neue Spalte heißt exakt wie die Spalte in Chemical_table
    df[prop_name] = df["Composition"].apply(
        lambda c: build_property_dict(c, prop_map)
    )
```

```python
def parse_formula(formula: str):
    """
    Zerlegt chemische Formeln wie:
      AlCl3 → {'Al': 1, 'Cl': 3}
      CaCl2-LaCl3 → {'Ca': 1, 'Cl': 5, 'La': 1}
    """
    pattern = r"([A-Z][a-z]*)(\d*)"
    parts = re.findall(pattern, str(formula))

    result = {}
    for elem, count in parts:
        count = int(count) if count else 1
        result[elem] = result.get(elem, 0) + count
    return result

print("Test parse_formula:", parse_formula("AlCl3"))

```

```python

def compute_atomic_radius_components(system_formula: str, radius_map: dict) -> float:
    elems = parse_formula(system_formula)
    total = 0.0
    for symbol, count in elems.items():
        if symbol in radius_map and pd.notna(radius_map[symbol]):
            total += radius_map[symbol] * count
    return total

# Die System-Spalte robust finden
system_col = [c for c in df.columns if "System" in c][0]
print("Verwende System-Spalte:", system_col)


df["Atomic radius(components)"] = df[system_col].apply(
    lambda s: compute_atomic_radius_components(s, atomic_radius_map)
)
```

### Ionische Polarizierbarkeit eines Elements

$$
\alpha_{\text{ion}} 
= \alpha_{\text{atom}}
\left( \frac{Z}{Z + q} \right)^2
$$

### Stöchiometrisch gewichtete gesamte ionische Polarizierbarkeit

$$
\alpha_{\text{sum}} 
= \sum_{i} n_i \, \alpha_{\text{ion},i}
$$

### Mittlere ionische Polarizierbarkeit pro Atom

$$
\alpha_{\text{mean}} 
= \frac{\alpha_{\text{sum}}}{\sum_i n_i}
$$


```python
def ensure_dict(x):
    """Spalten, die als String gespeicherte Dicts enthalten, in echte Dicts umwandeln."""
    if isinstance(x, dict):
        return x
    if isinstance(x, str):
        try:
            return ast.literal_eval(x)
        except Exception:
            return {}
    return {}


def parse_charge(charge_str):
    """
    '3+' -> +3
    '1-' -> -1
    Sonst 0
    """
    if not isinstance(charge_str, str):
        return 0
    s = charge_str.strip()
    if not s:
        return 0
    sign = -1 if s.endswith("-") else 1
    mag = s[:-1]
    try:
        mag = int(mag)
    except ValueError:
        return 0
    return sign * mag


def get_Z(symbol: str):
    """Nur Ordnungszahl Z aus dem Periodensystem holen."""
    try:
        return pt.elements.symbol(symbol).number
    except Exception:
        return None


def compute_ionic_polarizability_row(row):
    """
    Für eine df-Zeile:
      - berechne ionische Polarizierbarkeit je Element
      - aggregiere zu Summe & Mittelwert (stoichiometrisch gewichtet)
    """
    system_formula = row["System"]
    comp = parse_formula(system_formula)  # {'Al': 1, 'Cl': 3}

    pol_dict    = ensure_dict(row["polarizability_element[10-24Cm3]"])
    charge_dict = ensure_dict(row["ionic_charge_element"])

    total = 0.0
    n_atoms = 0

    for sym, stoich in comp.items():
        alpha_atom = pol_dict.get(sym)
        charge_str = charge_dict.get(sym)
        Z = get_Z(sym)

        if alpha_atom is None or Z is None or charge_str is None:
            continue

        q = parse_charge(charge_str)
        Z_eff_neutral = float(Z)
        Z_eff_ion = Z + q

        if Z_eff_ion <= 0:
            continue

        scale = (Z_eff_neutral / Z_eff_ion) ** 2
        alpha_ion = alpha_atom * scale

        total   += alpha_ion * stoich
        n_atoms += stoich

    sum_val  = total
    mean_val = total / n_atoms if n_atoms > 0 else math.nan
    return pd.Series([sum_val, mean_val])


df[["ionic_polarizability_sum_components",
    "ionic_polarizability_mean_components"]] = df.apply(
        compute_ionic_polarizability_row,
        axis=1
    )


print(df[["System",
          "ionic_polarizability_sum_components",
          "ionic_polarizability_mean_components"]].head())
```
