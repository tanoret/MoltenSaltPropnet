```python

import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler

numeric_df = df.select_dtypes(include='number')
scaler = StandardScaler()
normalized_data = scaler.fit_transform(numeric_df)
normalized_df = pd.DataFrame(normalized_data, columns=numeric_df.columns)
correlation_matrix = normalized_df.corr()

plt.figure(figsize=(6, 6))
sns.heatmap(correlation_matrix, 
            annot=True, 
            fmt=".2f", 
            cmap='coolwarm', 
            center=0, 
            square=True, 
            linewidths=0.5, 
            cbar_kws={"shrink": 0.75})
plt.title("Normalized Correlation Matrix Heatmap")
plt.tight_layout()
plt.show()

```

```python

import pandas as pd
import numpy as np

def parse_mol_frac(val):
    if val == "Pure Salt":
        return [1.0, 0.0, 0.0], True
    parts = [float(x) for x in val.split('-')]
    
    while len(parts) < 3:
        parts.append(0.0)
    return parts[:3], False

mol_frac_data = df['Mol Frac'].apply(parse_mol_frac)
df[['mol_frac_0', 'mol_frac_1', 'mol_frac_2']] = pd.DataFrame(mol_frac_data.apply(lambda x: x[0]).tolist(), index=df.index)
df['is_pure_salt'] = mol_frac_data.apply(lambda x: x[1])
```

```python
"""To understand what do we want to do with the missing data, is there something 
we can drop or is there anything which is better than expected,
FAZIT: No there is no full row, the best complete values has RHO
muB and Boil are horrible in terms of data complete"""

#the other values we just have are they at least complete
Mass_all_present = df[['Mol Mass']].notna().all(axis=1)
Mol_all_present = df[['Mol Frac']].notna().all(axis=1)
print('Complete: ', Mass_all_present.sum())
print("Mol", Mol_all_present.sum())

#RHO
rho_all_present = df[['rho_a', 'rho_b']].notna().all(axis=1)
rho_all_missing = df[['rho_a', 'rho_b']].isna().all(axis=1)
rho_partial = df[['rho_a', 'rho_b']].notna().any(axis=1) & ~rho_all_present
print("RHO:")
print("Complete rho:", rho_all_present.sum())
print("Only one value:", rho_partial.sum())
print("NAN", rho_all_missing.sum())

# muA
muA_all_present = df[['mu1_a', 'mu1_b']].notna().all(axis=1)
muA_all_missing = df[['mu1_a', 'mu1_b']].isna().all(axis=1)
muA_partial = df[['mu1_a', 'mu1_b']].notna().any(axis=1) & ~muA_all_present
print("\nmuA:")
print("Complete muA", muA_all_present.sum())
print("Only one value:", muA_partial.sum())
print("Complete NAN", muA_all_missing.sum())

# muB
muB_all_present = df[['mu2_a', 'mu2_b', 'mu2_c']].notna().all(axis=1)
muB_all_missing = df[['mu2_a', 'mu2_b', 'mu2_c']].isna().all(axis=1)
muB_partial = df[['mu2_a', 'mu2_b', 'mu2_c']].notna().any(axis=1) & ~muB_all_present

print("\nmuB:")
print("Complete muB:", muB_all_present.sum())
print("Only one:", muB_partial.sum())
print("NAN:", muB_all_missing.sum())

df['Melt(K)'] = pd.to_numeric(df['Melt(K)'], errors='coerce')
df['Boil(K)'] = pd.to_numeric(df['Boil(K)'], errors='coerce')
print("\nMelt(K):")
print("Available:", df['Melt(K)'].notna().sum())
print("Available:", df['Melt(K)'].isna().sum())

print("\nBoil(K):")
print("Available:", df['Boil(K)'].notna().sum())
print("Missing:", df['Boil(K)'].isna().sum())

# Checking if there exist a row which is fully complete and if yes which
rho_cols = ['rho_a', 'rho_b']
muA_cols = ['mu1_a', 'mu1_b']
muB_cols = ['mu2_a', 'mu2_b', 'mu2_c']
melt_col = 'Melt(K)'
boil_col = 'Boil(K)'
rho_complete = df[rho_cols].notna().all(axis=1)
muA_complete = df[muA_cols].notna().all(axis=1)
muB_complete = df[muB_cols].notna().all(axis=1)
melt_present = df[melt_col].notna()
boil_present = df[boil_col].notna()
all_complete = rho_complete & muA_complete & muB_complete & melt_present & boil_present
missing_muB = rho_complete & muA_complete & ~muB_complete & melt_present & boil_present
missing_muA = rho_complete & ~muA_complete & muB_complete & melt_present & boil_present
missing_muA_muB = rho_complete & ~muA_complete & ~muB_complete & melt_present & boil_present

print("\nDo we even have a complete row?")
print("Rows with ALL values complete:", all_complete.sum())
print("Rows complete EXCEPT muB:", missing_muB.sum())
print("Rows complete EXCEPT muA:", missing_muA.sum())
print("Rows complete EXCEPT muA and muB:", missing_muA_muB.sum())
```
